```python

import os
import geopandas as gpd


# ============================================================
# INPUT / OUTPUT
# ============================================================

INPUT_GPKG = r"C:\Users\Standard\Downloads\nepal-260826-free.gpkg\nepal.gpkg"

OUTPUT_DIR = r"F:\Jintu_RS_2\NERDRR\August 2026\Nepal Flood\OSM_Buildings"


# ============================================================
# MAIN CONVERSION FUNCTION
# ============================================================

def convert_gpkg_to_shp():

    print("=" * 70)
    print("GeoPackage → Shapefile Conversion")
    print("=" * 70)

    # --------------------------------------------------------
    # Check input
    # --------------------------------------------------------

    if not os.path.exists(INPUT_GPKG):
        print("\nERROR: Input GeoPackage not found!")
        print(INPUT_GPKG)
        return

    # --------------------------------------------------------
    # Create output directory
    # --------------------------------------------------------

    os.makedirs(OUTPUT_DIR, exist_ok=True)

    print("\nInput GeoPackage:")
    print(INPUT_GPKG)

    print("\nOutput directory:")
    print(OUTPUT_DIR)

    # --------------------------------------------------------
    # Find all layers
    # --------------------------------------------------------

    print("\nReading GeoPackage layers...")

    try:
        layers = gpd.list_layers(INPUT_GPKG)
    except Exception as e:
        print("\nERROR while reading GeoPackage:")
        print(e)
        return

    if layers.empty:
        print("\nNo layers found in the GeoPackage.")
        return

    print("\nAvailable layers:")
    print("-" * 70)

    for i, row in layers.iterrows():
        print(
            f"{i + 1}. "
            f"{row['name']} "
            f"({row['geometry_type']})"
        )

    print("-" * 70)

    # --------------------------------------------------------
    # Convert each layer
    # --------------------------------------------------------

    successful = 0
    failed = 0

    for i, row in layers.iterrows():

        layer_name = row["name"]

        print("\n" + "=" * 70)
        print(f"Processing layer: {layer_name}")
        print("=" * 70)

        try:

            # Read layer
            gdf = gpd.read_file(
                INPUT_GPKG,
                layer=layer_name
            )

            print(f"Features: {len(gdf)}")
            print(f"CRS: {gdf.crs}")

            # ------------------------------------------------
            # Clean filename
            # ------------------------------------------------

            safe_name = str(layer_name)

            invalid_chars = '<>:"/\\|?*'

            for char in invalid_chars:
                safe_name = safe_name.replace(char, "_")

            safe_name = safe_name.strip()

            if not safe_name:
                safe_name = f"layer_{i + 1}"

            # ------------------------------------------------
            # Output path
            # ------------------------------------------------

            output_shp = os.path.join(
                OUTPUT_DIR,
                safe_name + ".shp"
            )

            # ------------------------------------------------
            # Handle empty layer
            # ------------------------------------------------

            if gdf.empty:
                print("WARNING: Layer is empty.")
                print("Skipping...")
                continue

            # ------------------------------------------------
            # Export
            # ------------------------------------------------

            print("\nExporting:")

            gdf.to_file(
                output_shp,
                driver="ESRI Shapefile",
                encoding="UTF-8"
            )

            print("SUCCESS!")
            print(output_shp)

            successful += 1

        except Exception as e:

            failed += 1

            print("\nERROR converting layer:")
            print(layer_name)

            print("Error:")
            print(e)

    # ========================================================
    # SUMMARY
    # ========================================================

    print("\n")
    print("=" * 70)
    print("CONVERSION COMPLETED")
    print("=" * 70)

    print(f"Successful layers : {successful}")
    print(f"Failed layers     : {failed}")

    print("\nOutput folder:")
    print(OUTPUT_DIR)

    print("\nGenerated files:")

    for file in os.listdir(OUTPUT_DIR):

        if file.lower().endswith(".shp"):
            print("  ", file)

    print("\n" + "=" * 70)


# ============================================================
# RUN
# ============================================================

if __name__ == "__main__":
    convert_gpkg_to_shp()

```
